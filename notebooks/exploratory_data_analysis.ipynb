# Exploratory data analysis (SYNTHETIC data)
Run from the project root after `python -m src.data.generate_dataset`.

```python
import pandas as pd, numpy as np, matplotlib.pyplot as plt, seaborn as sns
df = pd.read_csv('../data/processed/football_injury_dataset.csv')
print(df.shape, df['injury_next_7_days'].mean())
df.iloc[:, :25].head()
```

```python
df.groupby('position')['injury_next_7_days'].agg(['mean','size'])
```

```python
cols = ['acute_chronic_workload_ratio','workload_monotony','workload_strain','previous_injury_count','strength_asymmetry_percent','age']
fig, ax = plt.subplots(2, 3, figsize=(13, 6))
for a, c in zip(ax.ravel(), cols):
    sns.kdeplot(data=df, x=c, hue='injury_next_7_days', common_norm=False, ax=a)
plt.tight_layout()
```

```python
sns.heatmap(df[cols + ['injury_next_7_days']].corr(method='spearman'), annot=True, fmt='.2f', cmap='vlag')
plt.title('Spearman correlations (association within synthetic data, not causation)')
```
